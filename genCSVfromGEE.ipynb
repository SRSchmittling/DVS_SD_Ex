This Notebook will generate a final .csv file

```python
#   Import Statements
import pandas as pd
import numpy as np
import os

all_files = os.listdir()
csv_files = [f for f in all_files if f.endswith('.csv')]
csv_files.sort()
print(f'Number of files: {len(csv_files)}')
```

```python
df = pd.DataFrame()
nfiles = 0
for csv_file in csv_files:
    print(f'Processing file: {csv_file}')
    nfiles += 1
    dfIn = pd.read_csv(csv_file)
    # Calculate a mean for each scenario
    # ssp245
    ssp245_cols = [col for col in dfIn.columns if col.startswith('ssp245_')]
    dfIn['mean_ssp245'] = dfIn[ssp245_cols].mean(axis=1)
    # ssp585
    ssp585_cols = [col for col in dfIn.columns if col.startswith('ssp585_')]
    dfIn['mean_ssp585'] = dfIn[ssp585_cols].mean(axis=1)
    df = pd.concat([df, dfIn], ignore_index=True)

print(f'Number of files processed: {nfiles}')
df.head(5)
```

```python
# Create an empty dataframe to store the data from
df = pd.DataFrame()
# Initialize a variable to track how many csv files processed.
nfiles = 0

# Process every csv_file in the list of csv_files
for csv_file in csv_files:
    print(f'Processing file: {csv_file}')

    # Increment counter to track how many csv files are processed
    nfiles += 1

    # Read in data from current csv_file
    dfIn = pd.read_csv(csv_file)

    # Calculate a mean for each scenario

    # ssp245
    ssp245_cols = [col for col in dfIn.columns if col.startswith('ssp245_')]
    dfIn['ssp245'] = dfIn[ssp245_cols].mean(axis=1)

    # ssp585
    ssp585_cols = [col for col in dfIn.columns if col.startswith('ssp585_')]
    dfIn['ssp585'] = dfIn[ssp585_cols].mean(axis=1)

    # Add the data from the current csv file to
    df = pd.concat([df, dfIn], ignore_index=True)

print(f'Number of files processed: {nfiles}')
```

```python
#   Remove Model values and extract means
df_means = df[['isodate','district','pcode','band','ssp245', 'ssp585']]
df_melted = df_means.melt(id_vars=['isodate','district','pcode','band'],
                    value_vars = ['ssp245','ssp585'],
                    var_name = 'scenario',
                    value_name = 'mean')

#   Print out first 10 rows
print(df_melted.head(10))
```

```python
#   Get counts of some of the fields for sanity check

# Get Earliest and Latest Date Strings
startDate = pd.to_datetime(df_melted['isodate']).min()
endDate = pd.to_datetime(df_melted['isodate']).max()
totalDays = (endDate-startDate).days+1


print(f"Date From: {startDate}")
print(f"Date To: {endDate}")
print(f"Total Days: {totalDays}")

# each scenario should have num districts * num days * num bands
print(df_melted['scenario'].value_counts())

# each district should have num days * num bands * num scenarios
print(df_melted['district'].value_counts())

# each band should have num districts * num days * num scenarios
print(df_melted['band'].value_counts())
```

```python
#   Create wide data (Date, District, PCode, Band, mean for district over all models)
df_final = df_melted.pivot(index=['isodate','pcode','district','scenario'], columns='band',values='mean')
print(df_final.head(10))
```
